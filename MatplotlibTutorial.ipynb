# Matplotlib Tutorial

## 1. Introduction to Matplotlib

Matplotlib is a Python library used for data visualization. It helps convert numerical data into meaningful graphs and charts. It works well with NumPy and Pandas and is widely used in data science and machine learning.

## 2. Installation & Import

```python
# Matplotlib comes pre-installed in Google Colab
# Run this only if needed
# !pip install matplotlib


import matplotlib.pyplot as plt
import numpy as np
```

```python
import matplotlib
matplotlib.__version__
```

## 3. Basic Line Plot

A line plot is used to show trends over time or ordered data.

```python
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]


plt.plot(x, y)
plt.show()
```

## 4. Line Plot with Labels and Title

Labels help make the graph understandable.

```python
plt.plot(x, y)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Simple Line Plot")
plt.show()
```

## 5. Multiple Line Plot

Used to compare multiple datasets.

```python
x = np.arange(1, 6)
y1 = x * 2
y2 = x * 3


plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")
plt.legend()
plt.show()
```

## 6. Bar Chart

Bar charts are used for comparing categorical data.

```python
categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]


plt.bar(categories, values)
plt.title("Bar Chart Example")
plt.show()
```

## 7. Horizontal Bar Chart

```python
plt.barh(categories, values)
plt.title("Horizontal Bar Chart")
plt.show()
```

## 8. Histogram

A histogram shows the distribution of data.

```python
data = np.random.randn(1000)


plt.hist(data, bins=20)
plt.title("Histogram")
plt.show()
```

## 9. Scatter Plot

Used to identify relationships between two variables.

```python
x = np.random.rand(50)
y = np.random.rand(50)


plt.scatter(x, y)
plt.title("Scatter Plot")
plt.show()
```

## 10. Pie Chart

Used to show proportions.

```python
labels = ['Python', 'Java', 'C', 'Others']
sizes = [40, 30, 20, 10]


plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

## 11. Figure Size

```python
plt.figure(figsize=(8, 4))
plt.plot(x, y)
plt.show()
```

## 12. Grid in Plot

```python
plt.plot(x, y)
plt.grid(True)
plt.show()
```

## 13. Subplots

Multiple plots in one figure.

```python
x = np.arange(1, 6)


plt.subplot(1, 2, 1)
plt.plot(x, x**2)
plt.title("Square")


plt.subplot(1, 2, 2)
plt.plot(x, x**3)
plt.title("Cube")


plt.show()
```

## 14. Using Matplotlib with Pandas

```python
import pandas as pd


df = pd.DataFrame({
'Year': [2019, 2020, 2021, 2022],
'Sales': [200, 250, 300, 350]
})


df.plot(x='Year', y='Sales', kind='line')
plt.show()
```

## 15. Saving a Plot

```python
x = np.arange(50)
y = np.random.rand(50)

plt.plot(x, y)
plt.savefig("plot.png")
plt.show()

#Matplotlib needs x and y with same number of values
#Here both have length 50, so it works
```

## 16. Advanced: Styling (Basic)

```python
plt.plot(x, y, linestyle='--', marker='o')
plt.title("Styled Plot")
plt.show()
```

## 17. Advanced: Real-time Style Use

```python
plt.style.use('ggplot')
plt.plot(x, y)
plt.show()
```

## 18. When to Use Which Plot

Line Plot → Trends

Bar Chart → Comparison

Histogram → Distribution

Scatter Plot → Relationship

Pie Chart → Proportion
