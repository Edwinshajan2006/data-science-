```python
#Q4 B
num_cols = df.select_dtypes(include=['int64','float64']).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())
```

```python
#Q4 C
cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)

print(df.isnull().sum())
```

```python
#6 a
avg_marks=df['marks'].mean()
avg_marks
```

```python
#6 b
above_avg = df[df['marks'] > avg_marks]
print(above_avg)
```

```python
#6 c
result = df[(df['attendance'] < 75) & (df['marks'] > avg_marks)]
result
```

```python
#Q 7 a
df.groupby('department').size()
```

```python
#q 7 b
dept_avg = df.groupby('department')['marks'].mean()
dept_avg
```

```python
#q 7 c
highest_dept = dept_avg.idxmax()
highest_dept
```

```python
#q 8
sorted_df = df.sort_values(by=['marks','attendance'],
                           ascending=[False,False])
sorted_df
```

```python
#q 9 a
df['result'] = df['marks'].apply(
    lambda x: 'Pass' if x >= 40 else 'Fail'
)

df[['marks','result']]
```

```python
#q 9 b
def performance(mark):
    if mark >= 80:
        return "Excellent"
    elif mark >= 50:
        return "Good"
    else:
        return "Needs Improvement"

df['performance'] = df['marks'].apply(performance)

(df[['marks','performance']])
```

```python
#q 9 c
df['performance'].value_counts()
```

```python
#q 10
df.reset_index(drop=True, inplace=True)
df.to_excel("cleaned_students_dataset.xlsx",
            index=False)

print("File exported successfully!")
from google.colab import files
files.download("cleaned_students_dataset.xlsx")
```
